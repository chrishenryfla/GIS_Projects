```python
from arcgis.features import FeatureLayer

layer_url = "https://services1.arcgis.com/jUJYIo9tSA7EHvfZ/arcgis/rest/services/DSPACE_INSPECTIONS_AUX_LAC_sync_script/FeatureServer"
primary_structure_layer = FeatureLayer(layer_url)
```

```python
# Print all fields/columns
print("Columns in the layer:")
for field in primary_structure_layer.properties.fields:
    print(f"- {field['name']} ({field['type']})")
```

```python
# Confirm exact field names
status_field = "PROPERTYSTATUS"  
battid_field = "BATTID"

# List of statuses to count
status_list = ["Compliant", "2nd Non Compliant", "1st Non Compliant", "Citation"]

# Query all features
features = primary_structure_layer.query()

# Convert to dataframe to pull unique values from
sdf = features.sdf

# Unique BATTID values excluding NA
battid_list = [val for val in sdf['BATTID'].unique() if str(val) != 'nan']

# Final output list to be filled
battid_out = []

# Loop through each battid
for battid in battid_list:
    entry = {"BATTID": battid}

    # Count each status
    for status in status_list:
        query = f"{status_field} = '{status}' AND {battid_field} = {battid}"
        try:
            count = primary_structure_layer.query(where=query, return_count_only=True)
        except:
            count = None
        entry[status] = count

    # Only keep entries where "Compliant" is a valid number and > 0
    if isinstance(entry["Compliant"], (int, float)) and entry["Compliant"] > 0:
        battid_out.append(entry)

# Print the resulting list of dictionaries
#for item in battid_out:
    #print(item)

df1 = pd.DataFrame(battid_out)
```

```python
# Confirm exact field names from layer properties
status_field = "PROPERTYSTATUS"  # replace with exact field name from layer
STANUM_field = "STANUM"          # numeric field

# List of statuses to count
status_list = ["Compliant", "2nd Non Compliant", "1st Non Compliant", "Citation"]

# Query all features (or filter if you want)
features = primary_structure_layer.query()

# Convert to Spatially Enabled DataFrame
sdf = features.sdf

#store unique values of STANUM
unique_values = sdf['STANUM'].unique()

# Filter out NA/None/NaN values
STANUM_list = [val for val in unique_values if str(val) != 'nan']

# Dictionary to hold results
results = {}

# Loop through each STANUM and count statuses
for STANUM in STANUM_list:
    STANUM_counts = {}
    for status in status_list:
        query = f"{status_field} = '{status}' AND {STANUM_field} = {STANUM}"
        try:
            count = primary_structure_layer.query(where=query, return_count_only=True)
        except Exception as e:
            print(f"Query failed for STANUM={STANUM}, STATUS={status}: {e}")
            count = None
        STANUM_counts[status] = count
    results[STANUM] = STANUM_counts

# Print results
for STANUM, counts in results.items():
    compliant_count = counts.get("Compliant")
    if compliant_count and compliant_count > 0:
        print(f"STANUM {STANUM}:")
        for status, count in counts.items():
            print(f"  {status}: {count}")
        print()
```

```python
print(df.columns.tolist())
```

```python
unique_values = df["INSPECTIONSTATUS"].unique()

print("Unique values in IPLPropStatus:")
for val in unique_values:
    print(val)
```

```python
# Filter only the statuses we care about
status_list = ["1st Non Compliant", "2nd Non Compliant"]
filtered_df = df[df["INSPECTIONSTATUS"].isin(status_list)]

# Group by GlobalID and collect unique statuses
status_counts = filtered_df.groupby("PARENTGUID")["INSPECTIONSTATUS"].unique()

# Keep only IDs that have BOTH statuses
ids_with_both = status_counts[status_counts.apply(lambda x: all(s in x for s in status_list))]

# Number of IDs with both
print(f"Number of GlobalIDs with both '1st' and '2nd Non Compliant': {len(ids_with_both)}")
```
