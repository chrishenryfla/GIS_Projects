```python
from PyPDF2 import PdfReader
import glob
import os
import pandas as pd

# -----------------------------------
# DIRECTORY
# -----------------------------------

folder = (
    r"C:\Users\e705941\OneDrive - County of Los Angeles\Defensible Space Unit - Compliance\2026 Compliance Document Creation\April"
)

# -----------------------------------
# FIND PDF FILES
# -----------------------------------

pdf_files = glob.glob(
    os.path.join(folder, "*.pdf")
)

pdf_files.sort()

# -----------------------------------
# GET PAGE COUNTS
# -----------------------------------

results = []

for file in pdf_files:

    try:

        reader = PdfReader(file)

        page_count = len(reader.pages)

        results.append({
            "file_name": os.path.basename(file),
            "page_count": page_count
        })

        print(
            f"{os.path.basename(file)}: "
            f"{page_count} pages"
        )

    except Exception as e:

        print(f"FAILED: {file}")
        print(e)

# -----------------------------------
# OPTIONAL DATAFRAME
# -----------------------------------

page_counts_df = pd.DataFrame(results)

print(page_counts_df)
```

```python
from pathlib import Path
from PyPDF2 import PdfReader, PdfMerger

# ==========================================
# PATHS
# ==========================================

input_folder = Path(
    r"C:\Users\e705941\OneDrive - County of Los Angeles\Defensible Space Unit - 2026 Defensible Space Cycle\Compliance\2026 Compliance Document Creation"
)

output_folder = input_folder / "merged_batches"
output_folder.mkdir(exist_ok=True)

# ==========================================
# SETTINGS
# ==========================================

TARGET_PAGES = 5000

# ==========================================
# GET PDF PAGE COUNTS
# ==========================================

pdf_files = sorted(input_folder.glob("*.pdf"))

pdf_info = []

for pdf in pdf_files:

    try:
        reader = PdfReader(str(pdf))
        page_count = len(reader.pages)

        pdf_info.append({
            "path": pdf,
            "pages": page_count
        })

        print(f"{pdf.name}: {page_count} pages")

    except Exception as e:
        print(f"Skipping {pdf.name}: {e}")

# ==========================================
# CREATE BATCHES BASED ON PAGE COUNTS
# ==========================================

batches = []
current_batch = []
current_pages = 0

for pdf in pdf_info:

    pdf_pages = pdf["pages"]

    # If adding this file exceeds target,
    # finalize current batch first
    if current_pages + pdf_pages > TARGET_PAGES and current_batch:

        batches.append(current_batch)

        current_batch = []
        current_pages = 0

    # Add file to batch
    current_batch.append(pdf)

    current_pages += pdf_pages

# Add remaining batch
if current_batch:
    batches.append(current_batch)

# ==========================================
# MERGE BATCHES
# ==========================================

for i, batch in enumerate(batches, start=1):

    merger = PdfMerger()

    batch_page_total = 0

    print(f"\nCreating Batch {i}")

    for pdf in batch:

        merger.append(str(pdf["path"]))

        batch_page_total += pdf["pages"]

        print(
            f"  Added: {pdf['path'].name} "
            f"({pdf['pages']} pages)"
        )

    output_pdf = output_folder / f"Merged_Batch_{i}.pdf"

    merger.write(str(output_pdf))
    merger.close()

    print(
        f"\nCreated: {output_pdf.name} "
        f"with {batch_page_total} pages"
    )
```
